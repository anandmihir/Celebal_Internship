# Week 8: Single Agent Pipeline Project

## Objective

Build a simple rule-based AI agent that routes user queries to the correct tool.


# Step 1: Import Required Libraries

We first import the required libraries.

- json → To print outputs in JSON format.
- re → To extract keywords using regular expressions.

```python
import json
import re
```

# Step 2: Create Calculator Tool

This tool evaluates mathematical expressions.

Example

Input:
calculate 25+5

Output:
30

```python
def calculator(expression):
    try:
        result = eval(expression, {"__builtins__": {}}, {})
        return result
    except Exception:
        raise ValueError("Invalid mathematical expression.")
```

# Step 3: Create Keyword Extraction Tool

This tool extracts meaningful keywords by

- converting text to lowercase
- removing punctuation
- removing stop words
- returning unique keywords

```python
def extract_keywords(text):

    stop_words = {
        "the","is","a","an","and","or","to","of","for",
        "with","this","that","in","on","it","are","was"
    }

    words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

    keywords = []

    for word in words:
        if word not in stop_words:
            keywords.append(word)

    return list(set(keywords))
```

# Step 4: Build the Agent Pipeline

The agent performs task routing.

Routing Rules

1. If query contains "calculate"
      → Calculator Tool

2. If query contains "keywords"
      → Keyword Extraction Tool

3. Otherwise
      → General Response

All outputs must follow the JSON format:

{
    "type": "...",
    "result": ...
}

```python
def agent(query):

    try:

        q = query.lower()

        # Route to Calculator
        if "calculate" in q:

            expression = q.replace("calculate", "").strip()

            answer = calculator(expression)

            return {
                "type": "calculation",
                "result": answer
            }

        # Route to Keyword Extractor
        elif "keywords" in q:

            text = q.replace("keywords", "").strip()

            answer = extract_keywords(text)

            return {
                "type": "keywords",
                "result": answer
            }

        # General Query
        else:

            return {
                "type": "general",
                "result": "General Response: " + query
            }

    except Exception as e:

        return {
            "type": "error",
            "result": str(e)
        }
```

# Step 5: Automated Validation

The following test cases verify

Calculator routing

Keyword routing

General routing

Error handling

```python
test_queries = [

    "calculate 15 + 20",

    "keywords Artificial Intelligence is changing healthcare",

    "Hello Agent",

    "calculate 10 +"

]

for query in test_queries:

    print("Input:", query)

    print(json.dumps(agent(query), indent=4))

    print("-"*50)
```

# Step 6: Interactive Agent

The user can continuously enter queries.

Type

exit

to stop the program.

```python
while True:

    query = input("Enter Query: ")

    if query.lower() == "exit":
        print("Agent Stopped.")
        break

    response = agent(query)

    print(json.dumps(response, indent=4))
```

# Conclusion

In this project, we successfully built a simple rule-based single agent pipeline that routes user queries to the appropriate tool based on predefined conditions.

The agent performs three main tasks:
- Routes mathematical queries containing **"calculate"** to the Calculator Tool.
- Routes queries containing **"keywords"** to the Keyword Extraction Tool.
- Handles all other queries using a General Response module.

The system also includes exception handling to safely process invalid inputs and returns every response in a structured JSON format with the keys **"type"** and **"result"**. Automated test cases and an interactive `while True` loop were used to validate the routing logic and ensure reliable performance.

This project demonstrates the basic concepts of agent pipelines, conditional task routing, tool integration, structured outputs, and error handling, forming a strong foundation for building more advanced AI agent systems.
